```python
sensor_features = [
    "CO2_Room",
    "H2_Room",
    "PM05_Room",
    "PM100_Room",
    "PM10_Room",
    "PM25_Room",
    "PM40_Room",
    "PM_Room_Typical_Size",
    "PM_Total_Room",
    "VOC_Room_RAW",
    "Temperature_Room",
    "Humidity_Room",
    "CO_Room"
]

sensor_features
```

```python
X = df[sensor_features]
```

```python
X.shape
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
X_scaled[:5]
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

inertia = []

for k in range(2, 11):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)
```

```python
plt.plot(range(2, 11), inertia, marker="o")
plt.title("Elbow Method")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.show()
```

```python
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)

kmeans.fit(X_scaled)
```

```python
cluster_labels = kmeans.labels_
```

```python
from sklearn.metrics import silhouette_score

sample_size = 10000

score = silhouette_score(
    X_scaled[:sample_size],
    cluster_labels[:sample_size]
)

print("Silhouette Score:", score)
```

```python
import pandas as pd

cluster_counts = pd.Series(cluster_labels).value_counts().sort_index()
cluster_counts
```

```python
df["cluster"] = cluster_labels

df[df["cluster"] == 2][sensor_features]
```

```python
cluster_profile = df.groupby("cluster")[sensor_features].mean()
cluster_profile
```

```python
sample_indices = df.sample(n=2000, random_state=42).index

X_hierarchical = X_scaled[sample_indices]
```

```python
from scipy.cluster.hierarchy import dendrogram, linkage

Z = linkage(X_hierarchical, method="ward")

plt.figure(figsize=(12, 6))
dendrogram(Z, truncate_mode="lastp", p=30)
plt.title("Hierarchical Clustering Dendrogram")
plt.xlabel("Clusters / Groups")
plt.ylabel("Distance")
plt.show()
```

```python
from scipy.cluster.hierarchy import fcluster

hierarchical_labels = fcluster(
    Z,
    t=4,
    criterion="maxclust"
)

hierarchical_labels
```

```python
pd.Series(hierarchical_labels).value_counts().sort_index()
```

```python
X_dbscan = X_hierarchical.copy()
```

```python
from sklearn.cluster import DBSCAN

dbscan = DBSCAN(eps=0.5, min_samples=10)

dbscan_labels = dbscan.fit_predict(X_dbscan)

pd.Series(dbscan_labels).value_counts().sort_index()
```

```python
from sklearn.neighbors import NearestNeighbors
import numpy as np
import matplotlib.pyplot as plt

neighbors = NearestNeighbors(n_neighbors=10)
neighbors.fit(X_dbscan)

distances, indices = neighbors.kneighbors(X_dbscan)

k_distances = np.sort(distances[:, 9])

plt.figure(figsize=(10, 5))
plt.plot(k_distances)
plt.title("K-Distance Graph for DBSCAN")
plt.xlabel("Observations")
plt.ylabel("10th Nearest Neighbor Distance")
plt.show()
```

```python
dbscan = DBSCAN(eps=5, min_samples=10)

dbscan_labels = dbscan.fit_predict(X_dbscan)

pd.Series(dbscan_labels).value_counts().sort_index()
```

```python
dbscan = DBSCAN(eps=1, min_samples=10)

dbscan_labels = dbscan.fit_predict(X_dbscan)

non_noise = dbscan_labels != -1

dbscan_silhouette = silhouette_score(
    X_dbscan[non_noise],
    dbscan_labels[non_noise]
)

print("DBSCAN Silhouette Score:", dbscan_silhouette)
```

```python
hierarchical_silhouette = silhouette_score(
    X_hierarchical,
    hierarchical_labels
)

print("Hierarchical Silhouette Score:", hierarchical_silhouette)
```

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)

print("Original features:", X_scaled.shape[1])
print("PCA components:", X_pca.shape[1])
```

## Final Conclusion

This project analyzed industrial multi-sensor data using unsupervised learning techniques to discover different sensor behaviour patterns.

After preprocessing and feature scaling, K-Means, Hierarchical Clustering, and DBSCAN were applied to the sensor data.

The Elbow Method suggested 4 clusters for K-Means. Among the tested clustering approaches, K-Means achieved the highest Silhouette Score of approximately 0.4822, indicating better cluster separation.

Cluster analysis showed different particulate-matter behaviour patterns. In particular, Cluster 2 contained only 6 observations with extremely high particulate-matter readings, making it an important group for further investigation.

DBSCAN also identified noise observations, which can be useful for detecting isolated or unusual sensor behaviour.

Overall, the project demonstrates how unsupervised learning can be used to discover patterns in industrial sensor data and support further investigation of unusual operating conditions.
